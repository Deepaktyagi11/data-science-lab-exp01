## Question 25

Read a copied table into a DataFrame using pd.read_clipboard(). Display its column names and then create a second DataFrame containing any three selected columns.

## Question 30

Copy a table with at least four columns from any webpage or spreadsheet and read it into lab_df. Write one sequence of Pandas statements to: display the DataFrame, display its column names, retrieve one single-word column, retrieve one column whose name contains a space, construct a smaller DataFrame from selected columns, add a new whole column, and finally delete that column.

```python

try:
    lab_df = pd.read_clipboard()
except Exception:
    print("Clipboard unavailable; using a sample table.")
    lab_df = pd.DataFrame({
        "Student": ["Aman", "Riya"],
        "Total Marks": [85, 92],
        "City": ["Mathura", "Delhi"],
        "Course": ["DS", "CSE"]
    })

display(lab_df)
print(lab_df.columns)
single_word_col = lab_df.columns[0]
display(lab_df[single_word_col])
space_columns = [col for col in lab_df.columns if " " in str(col)]
if space_columns:
    display(lab_df[space_columns[0]])
else:
    print("No column name containing a space was found in the copied table.")
selected_cols = list(lab_df.columns[:2])
small_lab_df = lab_df[selected_cols]
display(small_lab_df)

lab_df["NewColumn"] = "Practice"
display(lab_df)

lab_df = lab_df.drop(columns=["NewColumn"])
display(lab_df)
```
