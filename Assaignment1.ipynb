```python
def my_abs(val):
    if val < 0:
        return -val
    return val

print(my_abs(5))
print(my_abs(-8))
print(my_abs(0))
```

```python
def print_abs(val):
    if val < 0:
        print(-val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)
```

```python
def inc_val(val):
    val = val + 1

x = 7
inc_val(x)
print(x)
```

```python
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside Function:", val1, val2)

x = 6
y = 3

swap_message(x, y)

print("Outside Function:", x, y)
```

```python
counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
