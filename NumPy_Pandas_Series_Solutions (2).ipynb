### Q25.

```python
arr = np.array([[2, 4, 6, 8],
                [1, 3, 5, 7],
                [10, 20, 30, 40]])
print("Row-wise    var:", arr.var(axis=1))
print("Row-wise    std:", arr.std(axis=1))
print("Column-wise var:", arr.var(axis=0))
print("Column-wise std:", arr.std(axis=0))
```

**axis explanation:**
- `axis=0` → rows ke **upar se neeche** operate karta hai, matlab **har column** ka result milta hai.
- `axis=1` → columns ke **left se right** operate karta hai, matlab **har row** ka result milta hai.

### Q26.

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8],
              [9, 10],
              [11, 12]])
print("A @ B (2x2):\n", A @ B)

try:
    print(A * B)
except ValueError as e:
    print("Element-wise error:", e)
```

**Explanation:** Matrix multiplication me A ke columns (3) = B ke rows (3) hone chahiye, isliye result `2x2` aata hai.
Element-wise multiplication (`*`) me dono arrays ki shape **same** (ya broadcast-compatible) honi chahiye. `(2,3)` aur `(3,2)` compatible nahi hain, isliye **ValueError** aata hai.

### Q35.

```python
rng = np.random.default_rng(seed=42)
marks = rng.integers(20, 101, size=(12, 4))
print("Marks (rows = students, cols = tests):\n", marks)
student_avg = marks.mean(axis=1)
test_mean = marks.mean(axis=0)
test_std = marks.std(axis=0)
print("\nStudent averages:", np.round(student_avg, 2))
print("Test means      :", np.round(test_mean, 2))
print("Test std devs   :", np.round(test_std, 2))

passed_all = (marks >= 40).all(axis=1)
print("\nStudents passing all four tests:", passed_all.sum())
print("Their numbers:", (np.where(passed_all)[0] + 1).tolist())

high_avg = np.where(student_avg >= 75)[0] + 1
print("Students with average >= 75:", high_avg.tolist())

print("Test with greatest variation: Test", np.argmax(test_std) + 1,
      "(std =", round(test_std.max(), 2), ")")
```

### Q36.

```python
M = np.arange(21, 46).reshape(5, 5)
print("5x5 matrix:\n", M, "\nShape:", M.shape)

C = M[1:4, 1:4]
print("\nCentral 3x3:\n", C, "\nShape:", C.shape)

CT = C.T
print("\nTranspose:\n", CT, "\nShape:", CT.shape)

P = C @ CT
print("\nProduct C @ C.T:\n", P, "\nShape:", P.shape)

row_sums = P.sum(axis=1)
print("\nRow-wise sums:", row_sums, "\nShape:", row_sums.shape)

mx = P.max()
print("\nMaximum value:", mx)
print("Positions of max (row, col):", [tuple(p) for p in np.argwhere(P == mx)])
```

### Q37.

```python
rng = np.random.default_rng(7)
qty = rng.integers(5, 41, size=(6, 4))       # quantities sold
prices = np.array([150, 240, 90, 320])
print("Quantities (rows = days, cols = products):\n", qty)

revenue = qty * prices                       # broadcasting: har row * prices
daily_revenue = revenue.sum(axis=1)
product_revenue = revenue.sum(axis=0)
print("\nDaily revenue    :", daily_revenue)
print("Revenue / product:", product_revenue, " | Total:", product_revenue.sum())
print("Highest revenue day: Day", np.argmax(daily_revenue) + 1)

total_qty = qty.sum(axis=0)
print("\nTotal qty / product:", total_qty)
print("Best-selling product (by qty): Product", np.argmax(total_qty) + 1)

# 10% discount where total qty > 100 (naya array, original prices safe)
new_prices = np.where(total_qty > 100, prices * 0.9, prices)
new_product_revenue = (qty * new_prices).sum(axis=0)
print("\nOriginal prices:", prices)
print("New prices     :", new_prices)
print("New revenue / product:", new_product_revenue)
print("New total revenue    :", new_product_revenue.sum(), "(old:", product_revenue.sum(), ")")
```

### Q38.

```python
rng = np.random.default_rng(seed=123)
data = rng.normal(loc=50, scale=12, size=100)

outside = (data < 30) | (data > 70)
print("Values outside 30-70:", outside.sum())

cleaned = np.where(data < 30, 30, np.where(data > 70, 70, data))

comparison = pd.DataFrame({
    'Mean':     [data.mean(), cleaned.mean()],
    'Variance': [data.var(),  cleaned.var()],
    'Std Dev':  [data.std(),  cleaned.std()],
    'Min':      [data.min(),  cleaned.min()],
    'Max':      [data.max(),  cleaned.max()],
}, index=['Original', 'Cleaned']).round(3)
print(comparison)
```

```python
branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})
total = branch_a + branch_b + branch_c
print("Ordinary addition:\n", total)
common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
print("\nCommon quarters:", common.tolist())
print("Combined earnings of common quarters:\n", total[common])
total_zero = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print("\nMissing as zero:\n", total_zero)
```

### Q40.

```python
scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']
s = pd.Series(scores, index=required)
print("Series (required order preserved):\n", s)

missing = s[s.isnull()].index.tolist()
print("\nMissing student records:", missing)
valid = s[s >= 70]
print("\nValid scores >= 70:\n", valid)
d = valid.to_dict()
print("\nDictionary:", d)
rebuilt = pd.Series(d)
print("Rebuilt:\n", rebuilt)

print("\nLabelled values match:", rebuilt.equals(valid))
print("Label-by-label check :", all(valid[k] == rebuilt[k] for k in valid.index))
```
